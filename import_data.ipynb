```python
import sedia_api_fetchers
import pandas as pd
from sedia_api_fetchers.EUFT_retrieve_projects import SEDIA_GET_PROJECTS

```

```python
fetcher = SEDIA_GET_PROJECTS(
    flatten_metadata=True,      # Flatten nested JSON structures
    enrich_with_details=False   # Fetch detailed info (projects only)
)
```

```python
h2020_data = fetcher.get('h2020', save=True)
```
